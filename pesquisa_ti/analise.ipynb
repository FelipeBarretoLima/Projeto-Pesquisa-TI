```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("dados.csv")
df.head()
```

```python
df.info()
```

```python
df.columns
```

```python
df_tratado = df.copy()
df_tratado = df_tratado.drop('Carimbo de data/hora', axis=1)
df_tratado.head(5)
```

```python
df.columns
```

Renomeando colunas

```python
df_tratado = df_tratado.rename(columns={
    'Região onde mora:' : 'Região',
    'Gênero:' : 'Gênero', 
    'Escolaridade:' : "Escolaridade",
    'A formação é na área de tecnologia?': "Formado em tecnologia",
    'Tempo de atuação na área de tecnologia:': "Tempo de atuação",
    'Área de atuação que mais se adequa para sua realidade:': "Area de atuação",
    'Cargo que mais se adequa para a sua realidade:' : "Cargo",
    'Senioridade que se aplica para a sua realidade:' : 'Senioridade',
    'Faixa salarial mensal:': 'Faixa Salarial (mensal)',
    'Modalidade de trabalho que se aplica para a sua realidade:': "Modalidade de Trabalho",
    'Modalidade de atuação que se aplica para sua realidade:' : "Modalidade Atuação",
    'Segmento do local onde trabalha que mais se adequa as opções:':'Tipo empresa',
    'Perfil da empresa onde trabalha e tipo de alocação:' : "Perfil empresa",
    'Nível de satisfação com o mercado de trabalho:' :"Nivel satisfação"
})
df_tratado.head(5)
```

```python
df_tratado.info()
```

```python
df_tratado["Faixa Salarial (mensal)"] = df_tratado["Faixa Salarial (mensal)"].replace({
    "até 2 mil reais": "Até 2 mil",
    "entre 2 mil a 5 mil reais": "2 a 5 mil",
    "entre 5 mil a 8 mil reais": "5 a 8 mil",
    "entre 8 mil a 15 mil reais": "8 a 15 mil",
    "mais que 15 mil reais": "Mais de 15 mil",
    "Variável": "Variável"
})
```

```python
df_tratado['Formado em tecnologia'] = df_tratado["Formado em tecnologia"].replace(
    {
        "Sim":1,
        "Não":0
    }
    )
df_tratado['Formado em tecnologia'] = df_tratado['Formado em tecnologia'].astype(bool)

df_tratado.head(5)
```

```python
df_tratado['Ganha mais que 15 mil'] = df_tratado['Faixa Salarial (mensal)'] == "Mais de 15 mil"
df_tratado.head()
```

```python
df_tratado.columns
```

```python
plt.figure(figsize=(16,8))

sns.countplot(
    data=df_tratado,
    x="Escolaridade",
    hue="Ganha mais que 15 mil"
)
```

```python
plt.figure(figsize=(16, 6))

sns.countplot(
    data=df_tratado,
    x="Area de atuação",
    hue="Ganha mais que 15 mil"
)
```

```python
pd.crosstab(
    df_tratado["Senioridade"],
    df_tratado["Ganha mais que 15 mil"],
    normalize="index"
) * 100
```

```python
ordem = (
    df_tratado[df_tratado["Modalidade de Trabalho"] == "CLT"]
    ["Senioridade"]
    .value_counts()
    .index
)

plt.figure(figsize=(12, 6))

sns.countplot(
    data=df_tratado,
    x="Senioridade",
    hue="Modalidade de Trabalho",
    order=ordem
)

plt.title("Modalidade de Trabalho por Senioridade")
plt.xlabel("Senioridade")
plt.ylabel("Quantidade")
plt.legend(title="Modalidade de Trabalho")

plt.show()
```

```python
df_tratado['Senioridade'].value_counts()
```

```python
ordem_salario = [
    "Até 2 mil",
    "2 a 5 mil",
    "5 a 8 mil",
    "8 a 15 mil",
    "Mais de 15 mil",
    "Variável"
]

plt.figure(figsize=(16, 6))

sns.barplot(
    data=df_tratado,
    x="Faixa Salarial (mensal)",
    y="Nivel satisfação",
    order=ordem_salario
        )
```

```python
ordem_modalidade = [
    "Presencial",
    "Remoto",
    "Híbrido"
]

plt.figure(figsize=(16, 6))

sns.barplot(
    data=df_tratado,
    x="Modalidade Atuação",
    y="Nivel satisfação",
    order=ordem_modalidade,
    palette= "cividis"
        )

plt.title("Nível de Satisfação por Modelo de Trabalho por Modelo de Atuação")
```

```python
ordem = (
    df_tratado
    .groupby("Area de atuação")["Nivel satisfação"]
    .mean()
    .sort_values(ascending=False)
    .index
)

plt.figure(figsize=(16, 8))

sns.barplot(
    data=df_tratado,
    x="Area de atuação",
    y="Nivel satisfação",
    order=ordem,
    palette="cividis"
        )

plt.title("Média do nível de Satisfação por Área de atuação")
plt.show()
```

```python
df_tratado["Area de atuação"].value_counts()
```

```python
tabela_senioridade_salario = pd.crosstab(
    df_tratado['Senioridade'],
    df_tratado['Faixa Salarial (mensal)'],
    margins_name= "Total"
)
tabela_senioridade_salario
```

```python
ordem_salario = [
    "Até 2 mil",
    "2 a 5 mil",
    "5 a 8 mil",
    "8 a 15 mil",
    "Mais de 15 mil"
]

tabela_senioridade_salario = tabela_senioridade_salario[ordem_salario]

tabela_senioridade_salario.plot(
    kind="bar",
    stacked=True,
    colormap="viridis",
    figsize=(12, 6)
)

plt.title("Distribuição de Faixa Salarial por Senioridade")
plt.show()
```
